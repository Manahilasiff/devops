```python
import random

train = dataset["train"]

# Target numbers
BENIGN_TARGET = 80000
ATTACK_TARGET = 80000

# All 7 attack classes
attack_targets = {
    "DDoS": 11601,
    "DoS": 11601,
    "Mirai": 11601,
    "Spoofing": 11601,
    "Recon": 11601,
    "Web-based": 11601,
    "BruteForce": 10394
}

# Check total attack target
print("Attack target:", sum(attack_targets.values()))

# Get label and attack-class columns
labels = train["label"]
attack_class = train["attack_class"]

# Find indices for each class
class_indices = {}

print("\nFinding real record indices...")

# Benign
class_indices["Benign"] = [
    i for i, label in enumerate(labels)
    if label == 0
]

print("Benign available:", len(class_indices["Benign"]))

# Attack classes
for attack in attack_targets:
    class_indices[attack] = [
        i for i, name in enumerate(attack_class)
        if name == attack
    ]

    print(
        attack,
        "available:", len(class_indices[attack]),
        "needed:", attack_targets[attack]
    )

# Randomly select real records
selected_indices = []

# Benign
selected_indices.extend(
    random.sample(
        class_indices["Benign"],
        BENIGN_TARGET
    )
)

# Attack classes
for attack, target in attack_targets.items():

    selected_indices.extend(
        random.sample(
            class_indices[attack],
            target
        )
    )

# Shuffle the selected records
random.shuffle(selected_indices)

# Create final 160K training dataset
train_160k = train.select(selected_indices)

print("\n==============================")
print("160K TRAINING DATASET CREATED")
print("==============================")
print("Total records:", len(train_160k))
```

```python
import pandas as pd
import numpy as np

# Load the datasets
train_df = pd.read_csv("CIC_IoT_2023_train_160K.csv")
test_df = pd.read_csv("CIC_IoT_2023_test_40K.csv")

print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)

# Separate features and target
X_train = train_df.drop(columns=["attack_class", "label"])
y_train = train_df["label"]

X_test = test_df.drop(columns=["attack_class", "label"])
y_test = test_df["label"]

# Replace infinite values
X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_test = X_test.replace([np.inf, -np.inf], np.nan)

# Fill missing values using training-set medians
train_medians = X_train.median()

X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)

print("\nX_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test :", X_test.shape)
print("y_test :", y_test.shape)

print("\nTraining labels:")
print(y_train.value_counts())

print("\nTesting labels:")
print(y_test.value_counts())
```

**Phase 3 — Model Training**
Step 4: Train the Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

print("Training Random Forest...")

model.fit(X_train, y_train)

print("Training completed.")
```

**Phase 4 — Model Evaluation**
Step 5: Generate Predictions on the 40K Test Dataset

The trained model receives the 40K records it did not train on.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Predictions
y_pred = model.predict(X_test)

# Metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print("========== MODEL RESULTS ==========")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-Score : {f1:.4f}")

print("\n========== CLASSIFICATION REPORT ==========")
print(classification_report(y_test, y_pred, zero_division=0))

print("\n========== CONFUSION MATRIX ==========")
print(confusion_matrix(y_test, y_pred))
```
